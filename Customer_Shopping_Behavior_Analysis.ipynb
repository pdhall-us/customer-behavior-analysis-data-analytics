## Imputing null values with the median of each category

### Imputed with median because outliers can affect the mean

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median))
```

```python
df.isnull().sum()
```

## Renaming the columns according to snake casing as it will make the code consistent and improves readability.

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'puchase_amount'})
```

## Creating a new column age group

### It is really useful for marketing and understanding the behavior

```python
labels = ['Young Adult', 'Adult', 'Midddle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age', 'age_group']].head(10)
```

### Converting the frquency which is textual into numerical by creating new column purchase_frequency_days

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
